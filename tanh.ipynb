```python
import paddle
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches

plt.figure(figsize=(8,3))

# x = paddle.to_tensor([-10,10,1], dtype='float32', place=paddle.CPUPlace(), stop_gradient=False)
x= np.arange(-10,10,1)
x_exp = np.exp(x)
s = (x_exp - (1/x_exp))/(x_exp + (1/x_exp))

f = plt.subplot(121)
plt.plot(x, s, color='r')
# 添加文字说明
plt.text(-5., 0.9, r'$y=\sigma(x)$', fontsize=13)
# 设置坐标轴格式
currentAxis=plt.gca()
currentAxis.xaxis.set_label_text('x', fontsize=15)
currentAxis.yaxis.set_label_text('y', fontsize=15)

plt.show
```

```python
import numpy as np

# 生成10x10的随机数矩阵
p = np.random.randn(10, 10)

# 创建布尔掩码数组，大于0的位置为True，否则为False
q = (p > 0)

# 统计大于0的元素个数
positive_count = np.sum(q)

# 或者直接使用 p > 0 进行求和
# positive_count = np.sum(p > 0)

print("生成的随机矩阵：")
print(p)
print(f"\n大于0的元素个数：{positive_count}")
print(f"\n矩阵总元素数：{p.size}")

# 如果需要，可以查看q的数据类型和内容
print(f"\nq的数据类型：{type(q)}")
print(f"q的具体类型：{q.dtype}")
print("q的示例内容：")
print(q[:3, :3])  # 显示前3行3列作为示例
```
